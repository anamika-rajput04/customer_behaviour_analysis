```python
import pandas as pd
df=pd.read_csv('customer_shopping_behavior.csv')
```

```python
#Group the data by Category, find the median Review Rating for each category,
#replace missing ratings with that category's median, and save the updated values back into the Review Rating column.

df['Review Rating']=df.groupby('Category')['Review Rating'].transform(lambda x : x.fillna(x.median()))
```

```python
df.isnull().sum()
```

```python
df.columns=df.columns.str.lower()
df.columns=df.columns.str.replace(' ','_')
df=df.rename(columns={'purchase_amount_(usd)':'purchase_amount'})
```

```python
df.columns
```

```python
#create a column age_group

labels = ['Young Adult', 'Adult', 'Mid-aged', 'Senior']

df['age_group'] = pd.qcut(
    df['age'],
    q=4,
    labels=labels
)
```

```python
df[['age', 'age_group']].head(10)
```

```python
#create column purchase_frequency_days

frequency_mapping={
    'Fortnightly':14,
    'Weekly':7,
    'Monthly':30,
    'Quarterly':90,
    'Bi-Weekly':14,
    'Annually':365,
    'Every 3 Months':90,
}
df['purchase_frequency_days']=df['frequency_of_purchases'].map(frequency_mapping)
```

```python
df[['purchase_frequency_days','frequency_of_purchases']].head(10)
```

```python
df[['discount_applied', 'promo_code_used']].head(10)
```

```python
(df['discount_applied']==df['promo_code_used']).all()
```

```python
df=df.drop('promo_code_used',axis=1)
```

```python
df.columns
```

```python
from sqlalchemy import create_engine  #create engine helps to connect python to sql

username = 'root'
password = 'root'
host = '127.0.0.1'
port = '3306'
database = 'customer_behavior'

#Connect Python to my MySQL database customer_behavior using username root, password root, host 127.0.0.1, and port 3306
engine = create_engine(
    f'mysql+mysqlconnector://{username}:{password}@{host}:{port}/{database}'
)

table_name = "customer"

#Take this DataFrame and put it into a SQL database table
df.to_sql(
    table_name, #Put the data into a table called customer.
    engine, #Use the MySQL connection that we created above.
    if_exists="replace", #Python what to do if the customer table already exists Delete the old customer table and create it again with the new data.
    index=False
)

print(f"Data Successfully loaded into table '{table_name}' in database '{database}'.")
```
